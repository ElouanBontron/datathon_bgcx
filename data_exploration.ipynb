```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
folder = "data"
file = "champs_elysees_1.csv"

arc_id = 4264 # 4624 for champs elysees, XX for convention, XX for saints peres
```

```python
data = pd.read_csv(f"{folder}/{file}", encoding="utf-8", sep=";")
data
```

```python
columns_to_keep = ["Identifiant arc", "Date et heure de comptage", "Débit horaire", "Taux d'occupation", "Etat trafic", "Etat arc"]
```

```python
df = data[columns_to_keep]
df = df[df['Identifiant arc'] == arc_id]
df.drop(columns=['Identifiant arc'], inplace=True)
df.columns = ['datetime', 'flow', 'occupancy', 'traffic_state', 'arc_state']
df
```

```python
df = df[
    (df[["flow"]].notna().any(axis=1)) & 
    (df[["occupancy"]].notna().any(axis=1))
]
df
```

```python
df["traffic_state"] = pd.Categorical(
    df["traffic_state"],
    categories=['Fluide', 'Pré-saturé', 'Bloqué', 'Saturé'],
    ordered=True
)

df["open_arc"] = df["arc_state"] == "Ouvert"
```

```python
df["datetime"] = pd.to_datetime(
    df["datetime"],
    format="ISO8601"
)
```

```python
df.drop(columns=["arc_state"], inplace=True)
```

```python
df.reset_index(drop=True, inplace=True)
df
```
